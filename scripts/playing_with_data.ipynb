```python
import pandas as pd
import numpy as np
import pathlib as Path
import matplotlib.pyplot as plt

data = pd.read_csv("/Users/Dell_G3/Documents/Projects/Back-or-Sack-final-project-2026/data/processed/team_games.csv")
```

Home Advantage


```python
ppg = data.groupby("home")["points"].mean()
print(ppg)
```

Home advantage by season


```python
ppg_seasonal = data.groupby(['season', 'home'])['points'].mean()
print(ppg_seasonal)
```

```python
table = ppg_seasonal.unstack()

fig, ax = plt.subplots()

ax.plot(table.index, table[1], marker="o", label="Home")
ax.plot(table.index, table[0], marker="o", label="Away")

ax.set_title("Home advantage across seasons")
ax.set_xlabel("Season")
ax.set_ylabel("Points per game")
ax.set_ylim(0, 2)
ax.legend()
plt.show()
```
